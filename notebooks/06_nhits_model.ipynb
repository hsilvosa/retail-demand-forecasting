# N-HiTS probabilistic model
N-HiTS consumes 168 days of history and produces all 28 horizons directly. The Silver panel is mapped to NeuralForecast fields `unique_id`, `ds`, and `y`; state-specific SNAP is collapsed into one series-relevant flag. Known-future weekday, month, SNAP, planned price, and `price_missing` are supplied for every horizon.

A robust scaler is fitted inside each training fold by NeuralForecast, avoiding a global normalization fitted on future periods. Price null becomes zero only after retaining the missing-price flag. Raw demand is optimized with multi-quantile loss at q05/q50/q95, and outputs are clipped at zero. A recurrent one-step model was rejected because recursive error accumulation is harder to diagnose for inventory decisions.

The development profile uses only 100 optimization steps, so its score is a pipeline smoke result rather than a serious neural benchmark. The model uses CPU in development and CUDA in the optional full image. Interpret WRMSSE and interval coverage together; a narrow but poorly calibrated interval is not operationally useful.
